## An Engine class: the part that does the work

```python
class Engine:
    def start(self):
        print("Engine on")

Engine().start()
```

## Composition: a Car creates its own Engine

```python
class Car:
    def __init__(self):
        self.engine = Engine()
    def start(self):
        self.engine.start()

print(type(Car().engine).__name__)
```

## Using the Car: we only talk to the car

```python
c = Car()
c.start()
```
