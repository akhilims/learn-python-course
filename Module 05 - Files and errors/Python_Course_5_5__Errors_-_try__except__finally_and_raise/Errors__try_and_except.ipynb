## Catching errors with try and except

```python
try:
    n = int("abc")
except ValueError:
    print("Not a number")
```

## Reading the error message with as e

```python
try:
    print(10 / 0)
except ZeroDivisionError as e:
    print("Error:", e)
```

## Handling a missing file: FileNotFoundError

```python
try:
    open("missing.txt")
except FileNotFoundError:
    print("No such file")
```
