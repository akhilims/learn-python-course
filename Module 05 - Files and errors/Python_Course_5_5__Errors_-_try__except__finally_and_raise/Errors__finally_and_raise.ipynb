finally: code that always runs

```python
try:
    n = int("42")
except ValueError:
    print("bad")
finally:
    print("done")
```

raise: report your own error

```python
def set_age(age):
    if age < 0:
        msg = "age cannot be negative"
        raise ValueError(msg)
    return age
```

Catching your own error with try and except

```python
try:
    set_age(-1)
except ValueError as e:
    print(e)
```
