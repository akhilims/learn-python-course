# Python functions: def and call

```python
def greet(name):
    print("Hello,", name)
greet("Asha")
greet("Ben")
```

# return vs print in Python

```python
def add(a, b):
    return a + b
print(add(2, 3))
```

# Using the returned value

```python
total = add(4, 5)
print(total)
print(total * 2)
```

# A function that answers yes or no

```python
def is_even(n):
    return n % 2 == 0
print(is_even(10))
```
