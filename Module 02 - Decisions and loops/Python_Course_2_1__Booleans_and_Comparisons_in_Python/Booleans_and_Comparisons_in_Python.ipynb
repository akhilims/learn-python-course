Comparing numbers with comparison operators

```python
price = 40
budget = 50
print(price < budget)
print(price == budget)
print(price != budget)
print(price >= 45)
```

Comparing strings: case matters

```python
print("apple" == "apple")
print("apple" == "Apple")
```

Common mistake: one equals sign versus two

```python
print(price = budget)
```

Real-world check with and, or, not

```python
price = 40
budget = 50
in_stock = True
can_buy = price <= budget and in_stock
print(can_buy)
print(not in_stock)
print(price > 100 or in_stock)
```
