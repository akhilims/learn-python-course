# Setting up Rex again

```python
class Dog:
    pass
d = Dog()
d.name = "Rex"
print(d.name)
```

# A second object with its own name

```python
e = Dog()
e.name = "Bella"
print(e.name)
```

# A list of objects with a for loop

```python
pets = [d, e]
for p in pets:
    print(p.name)
```

# isinstance: was it made from this class?

```python
print(isinstance(d, Dog))
print(d.name == e.name)
```
