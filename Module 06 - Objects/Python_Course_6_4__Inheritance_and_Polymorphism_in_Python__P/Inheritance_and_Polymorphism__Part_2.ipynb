Setting up the Animal and Dog classes

```python
class Animal:
    def speak(self):
        print("...")
class Dog(Animal):
    def speak(self):
        print("Woof")
Dog().speak()
```

Adding a Cat class

```python
class Cat(Animal):
    def speak(self):
        print("Meow")
Cat().speak()
```

Polymorphism: one call, many answers

```python
for a in [Dog(), Cat()]:
    a.speak()
```

super(): build on the parent's method

```python
class Puppy(Dog):
    def speak(self):
        super().speak()
        print("Yip")
Puppy().speak()
```
