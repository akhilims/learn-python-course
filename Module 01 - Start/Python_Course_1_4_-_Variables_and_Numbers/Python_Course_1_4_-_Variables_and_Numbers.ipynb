# Variables and numbers

```python
# A variable is a labelled box
meal_price = 20
restaurant = "Luigi's Kitchen"
print(restaurant)
print(meal_price)
```

## Good names

```python
# Good names: lowercase, words joined by underscores
tip_percent = 10
number_of_guests = 4
table_2 = 55
# Bad names (they would fail):
# 2table = 55   starts with a digit
# my price = 5  has a space
print(tip_percent)
print(number_of_guests)
print(table_2)
```

## int and float

```python
whole = 7
decimal = 7.5
print(whole)
print(decimal)
# Division with / always gives a float
print(10 / 5)
print(10 / 4)
```

```python
a = 17
b = 5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
# Floor division, remainder and power
print(17 // 5)
print(17 % 5)
print(2 ** 10)
minutes = 135
print(minutes // 60)
print(minutes % 60)
```

```python
# A variable can change
score = 10
print(score)
score = 15
print(score)
score = score + 5
print(score)
```
