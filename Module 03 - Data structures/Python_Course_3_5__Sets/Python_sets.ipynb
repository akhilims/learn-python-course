Python sets: store unique items

```python
tickets = {7, 3, 7, 5}
print(tickets)
print(len(tickets))
```

Changing a set: add, remove, discard

```python
tickets.add(9)
tickets.add(7)
print(tickets)
```

Fast checks and removing duplicates

```python
print(7 in tickets)
nums = [1, 2, 2, 3, 3, 3]
print(set(nums))
```

Set operations: union and intersection

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a | b)
print(a & b)
```
