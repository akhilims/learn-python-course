## Defining __init__ and self in a class

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

## Creating an object: values go into __init__

```python
d = Dog("Rex", 3)
print(d.name, d.age)
```

## Changing an attribute later

```python
d.age = d.age + 1
print(d.age)
```
