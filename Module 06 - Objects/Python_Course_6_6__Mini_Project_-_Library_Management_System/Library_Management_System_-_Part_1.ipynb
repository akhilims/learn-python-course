## A Library class with a dictionary of books

```python
class Library:
    def __init__(self):
        self.books = {}
    def add(self, title):
        self.books[title] = False
```

## Create a library and add books

```python
lib = Library()
lib.add("Python")
lib.add("Maths")
print(lib.books)
```

## Show the library shelf with a for loop

```python
for title in lib.books:
    print(title, lib.books[title])
```
