## LEFT JOIN: find a state with no rows

```python
import sqlite3
import pandas as pd
con = sqlite3.connect("power.db")
sql = """
SELECT s.name, COUNT(g.gwh) AS n
FROM states s
LEFT JOIN generation g
ON g.state = s.code
GROUP BY s.name
HAVING n = 0
"""
print(pd.read_sql(sql, con))
```
