## Dataclass: the Book blueprint

```python
from dataclasses import dataclass
@dataclass
class Book:
    title: str
    pages: int
```

## Creating and printing a dataclass

```python
b = Book("Python", 300)
print(b)
```

## Equality and changing fields

```python
print(b == Book("Python", 300))
b.pages = 350
print(b.pages)
```
