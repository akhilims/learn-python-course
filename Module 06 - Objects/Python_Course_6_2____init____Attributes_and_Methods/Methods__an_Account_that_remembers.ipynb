Writing a method: the Account class

```python
class Account:
    def __init__(self):
        self.balance = 0
    def deposit(self, n):
        self.balance = self.balance + n
```

Calling a method with a dot

```python
acc = Account()
acc.deposit(50)
print(acc.balance)
```

The object remembers between calls

```python
acc.deposit(25)
print(acc.balance)
```
