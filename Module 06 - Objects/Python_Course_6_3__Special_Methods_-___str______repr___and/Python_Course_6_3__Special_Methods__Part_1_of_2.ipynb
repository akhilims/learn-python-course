A simple Book class

```python
class Book:
    def __init__(self, title):
        self.title = title
```

Printing an object without __str__

```python
b = Book("Python 101")
print(b)
```

Printing objects nicely with __str__

```python
class Book:
    def __init__(self, title):
        self.title = title
    def __str__(self):
        return self.title
print(Book("Python 101"))
```
