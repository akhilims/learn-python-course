Iterable: a list of lap times

```python
laps = [62, 58, 61]
```

iter() and next(): ask for one lap

```python
it = iter(laps)
print(next(it))
```

next() again: the iterator remembers

```python
print(next(it))
print(next(it))
```

The default value of next()

```python
print(next(it, "done"))
```

The for loop uses iter() and next()

```python
for lap in laps:
    print(lap)
```
