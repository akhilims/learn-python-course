Python recursion explained

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
```

RecursionError: a function with no base case

```python
def bad(n):
    return bad(n - 1)
print(bad(3))
```

Python docstrings

```python
def word_count(text):
    """Return how many words are in text."""
    return len(text.split())
print(word_count("to be or not"))
```
