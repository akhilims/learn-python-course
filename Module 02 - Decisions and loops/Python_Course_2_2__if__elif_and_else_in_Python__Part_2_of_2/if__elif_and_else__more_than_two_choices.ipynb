elif: a movie ticket price

```python
age = 67
if age < 12:
    price = 5
elif age >= 65:
    price = 6
else:
    price = 10
print(f"Ticket: {price} dollars")
```

Order matters: a grade from a score

```python
score = 82
if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 50:
    print("Grade C")
else:
    print("Try again")
```
