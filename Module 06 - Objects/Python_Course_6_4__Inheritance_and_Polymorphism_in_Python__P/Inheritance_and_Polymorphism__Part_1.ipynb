The parent class: a general Animal

```python
class Animal:
    def speak(self):
        print("...")
```

A child class inherits from its parent

```python
class Dog(Animal):
    pass
d = Dog()
d.speak()
```

Overriding a method in the child class

```python
class Dog(Animal):
    def speak(self):
        print("Woof")
Dog().speak()
```
