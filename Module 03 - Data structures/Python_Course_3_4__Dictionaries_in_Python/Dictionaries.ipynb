Python dictionaries: store key and value pairs

```python
ages = {"Asha": 21, "Ben": 25}
print(ages["Asha"])
ages["Cara"] = 30
print(ages)
```

KeyError: a missing key is an error

```python
print(ages["Zed"])
```

Dictionary get(): avoid KeyError

```python
print(ages.get("Zed", 0))
print("Asha" in ages)
del ages["Ben"]
```

Loop over a dictionary with items()

```python
for name, age in ages.items():
    print(name, age)
```
