Python lambda functions

```python
square = lambda n: n * n
print(square(5))
```

Python sorted with key

```python
words = ["pear", "fig", "banana"]
print(sorted(words))
print(sorted(words, key=len))
```

Python sorted with key and lambda

```python
people = [("Asha", 21), ("Ben", 19)]
print(sorted(people, key=lambda p: p[1]))
```
